```python
class Product:
    def __init__(self, product_id, name, description, price):
        self.product_id = product_id
        self.name = name
        self.description = description
        self.price = price
        self.active = True  # Product is active by default

    def update(self, name=None, description=None, price=None):
        if name:
            self.name = name
        if description:
            self.description = description
        if price:
            self.price = price
        print(f"Product {self.product_id} updated.")

    def suspend(self):
        self.active = False
        print(f"Product {self.product_id} suspended.")

    def reactivate(self):
        self.active = True
        print(f"Product {self.product_id} reactivated.")

    def get_details(self):
        return {
            "Product ID": self.product_id,
            "Name": self.name,
            "Description": self.description,
            "Price": self.price,
            "Status": "Active" if self.active else "Suspended"
        }
```

```python
# Create a product
prod = Product("PRD001", "Health Insurance", "Comprehensive health insurance plan", 1000)

# Update product details
prod.update(price=1200, description="Updated health insurance plan")

# Suspend the product
prod.suspend()

# Reactivate the product
prod.reactivate()

# Print product details
print(prod.get_details())
```
