```python
from datetime import datetime

class Payment:
    def __init__(self, amount, due_date):
        self.amount = amount
        self.due_date = due_date  # datetime object
        self.paid = False
        self.payment_date = None
        self.penalty = 0.0

    def process_payment(self):
        if not self.paid:
            self.paid = True
            self.payment_date = datetime.now()
            print(f"Payment of {self.amount} processed on {self.payment_date.strftime('%Y-%m-%d')}")
        else:
            print("Payment already processed.")

    def send_reminder(self):
        if not self.paid:
            days_remaining = (self.due_date - datetime.now()).days
            if days_remaining > 0:
                print(f"Reminder: Your payment of {self.amount} is due in {days_remaining} day(s).")
            elif days_remaining == 0:
                print(f"Reminder: Your payment of {self.amount} is due today!")
            else:
                print(f"Alert: Your payment of {self.amount} is overdue by {-days_remaining} day(s)!")

    def apply_penalty(self, daily_penalty_rate=0.05):
        if not self.paid:
            days_overdue = (datetime.now() - self.due_date).days
            if days_overdue > 0:
                self.penalty = self.amount * daily_penalty_rate * days_overdue
                print(f"Penalty of {self.penalty:.2f} applied for {days_overdue} day(s) overdue.")
            else:
                print("No penalty applied. Payment is not overdue.")
        else:
            print("No penalty applied. Payment is already made.")

    def get_summary(self):
        return {
            "Amount": self.amount,
            "Due Date": self.due_date.strftime("%Y-%m-%d"),
            "Paid": self.paid,
            "Payment Date": self.payment_date.strftime("%Y-%m-%d") if self.payment_date else None,
            "Penalty": self.penalty
        }
```

```python
from datetime import datetime, timedelta

# payment due 2 days from now
payment_due_date = datetime.now() + timedelta(days=2)
payment = Payment(300, payment_due_date)

payment.send_reminder()  # Reminder before due date

# payment overdue by adjusting due_date
payment.due_date = datetime.now() - timedelta(days=3)

payment.send_reminder()  # Overdue reminder

payment.apply_penalty()  # Apply penalty for overdue

payment.process_payment()  # Mark payment as processed

print(payment.get_summary())  
```
