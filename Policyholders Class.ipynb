```python
#1.Class Creation

class Policyholder:
    def __init__(self, name, policy_id):
        self.name = name
        self.policy_id = policy_id
        self.active = True
        self.payments = []

    def suspend(self):
        self.active = False

    def reactivate(self):
        self.active = True

    def add_payment(self, payment):
        self.payments.append(payment)

    def get_details(self):
        return {
            "Name": self.name,
            "Policy ID": self.policy_id,
            "Status": "Active" if self.active else "Suspended",
            "Payments": [p.get_summary() for p in self.payments]
        }

# Example usage:

# Since payments are referenced, create a mock Payment class for this test
class Payment:
    def __init__(self, amount):
        self.amount = amount

    def get_summary(self):
        return {"Amount": self.amount}

# Create a policyholder
holder = Policyholder("Ayobola", "PH1001")

# Create a payment and add it
payment = Payment(500)
holder.add_payment(payment)

# Print policyholder details
print(holder.get_details())
```
