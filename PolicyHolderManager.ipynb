```python
class Policyholder:
    def __init__(self, name, policy_id):
        self.name = name
        self.policy_id = policy_id
        self.active = True

    def suspend(self):
        self.active = False

    def reactivate(self):
        self.active = True

    def get_details(self):
        return {
            "Name": self.name,
            "Policy ID": self.policy_id,
            "Status": "Active" if self.active else "Suspended",
        }


class PolicyholderManager:
    def __init__(self):
        self.policyholders = {}  # policyholders with policy_id as key

    def register(self, name, policy_id):
        if policy_id in self.policyholders:
            print(f"Policyholder with ID {policy_id} already exists.")
        else:
            self.policyholders[policy_id] = Policyholder(name, policy_id)
            print(f"Policyholder {name} registered with ID {policy_id}.")

    def suspend(self, policy_id):
        if policy_id in self.policyholders:
            self.policyholders[policy_id].suspend()
            print(f"Policyholder {policy_id} suspended.")
        else:
            print(f"Policyholder {policy_id} not found.")

    def reactivate(self, policy_id):
        if policy_id in self.policyholders:
            self.policyholders[policy_id].reactivate()
            print(f"Policyholder {policy_id} reactivated.")
        else:
            print(f"Policyholder {policy_id} not found.")

    def get_policyholder_details(self, policy_id):
        if policy_id in self.policyholders:
            return self.policyholders[policy_id].get_details()
        else:
            return f"Policyholder {policy_id} not found."

    def list_all(self):
        return [ph.get_details() for ph in self.policyholders.values()]
```

```python
#Sample

manager = PolicyholderManager()

# two policyholders
manager.register("Ayobola", "PH1001")
manager.register("Tunde", "PH1002")

# Suspend one
manager.suspend("PH1002")

# Reactivate one
manager.reactivate("PH1002")

# Get details
print(manager.get_policyholder_details("PH1001"))
print(manager.get_policyholder_details("PH1002"))

# List all policyholders
print(manager.list_all())
```
